```python
df.drop(
    columns=[
        "Order_ID",
        "Customer_ID",
        "Product_ID",
        "Order_Amount",
        "Profit_Amount",
        "Profit_Margin_Percent",
        "Unit_Price",
        "Quantity",
        "Discount_Amount",
        "Shipping_Cost",
        "Tax_Amount",
        "Customer_Lifetime_Value"
    ],
    inplace=True
)
```

```python
X = df.drop("High_Value_Order", axis=1)
y =df["High_Value_Order"]
```

```python
categorical_cols = X.select_dtypes(include=["object"]).columns.tolist()
numerical_cols = X.select_dtypes(exclude=["object"]).columns.tolist()

print("Categorical Columns:", len(categorical_cols))
print("Numerical Columns", len(numerical_cols))
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_cols
        )
    ],
    remainder="passthrough"
)
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,          # use encoded labels
    test_size=0.20,
    stratify=y_encoded, # use encoded labels
    random_state=42
)
```

```python
print(y_train.dtype)
print(np.unique(y_train))
```

```python
rf_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),

        (
            "classifier",
            RandomForestClassifier(
                n_estimators=200,
                random_state=42
            )
        )
    ]
)
```

```python
rf_model.fit(X_train, y_train)
```

```python
y_pred = rf_model.predict(X_test)
```

```python
accuracy = accuracy_score(y_test, y_pred)

print(f"Accuracy: {accuracy:.4f}")
```

```python
cm = confusion_matrix(y_test, y_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=rf_model.classes_
)

disp.plot(cmap="Blues")

plt.title("Confusion Matrix")

plt.show()
```

#### Aggregated feature importance (correctly aggregated).

```python
import pandas as pd
import matplotlib.pyplot as plt

# Get feature names after preprocessing
feature_names = rf_model.named_steps["preprocessor"].get_feature_names_out()

# Get feature importances
importances = rf_model.named_steps["classifier"].feature_importances_

# Create DataFrame
fi = pd.DataFrame({
    "Feature": feature_names,
    "Importance": importances
})

# Function to recover original feature name
def get_original_feature(feature):
    # OneHotEncoded categorical features
    if feature.startswith("cat__"):
        return feature.replace("cat__", "").split("_")[0]

    # Numerical features (passed through)
    elif feature.startswith("remainder__"):
        return feature.replace("remainder__", "")

    return feature

# Aggregate importances
fi["Original_Feature"] = fi["Feature"].apply(get_original_feature)

feature_importance = (
    fi.groupby("Original_Feature", as_index=False)["Importance"]
      .sum()
      .sort_values(by="Importance", ascending=False)
)

# Display
display(feature_importance)

# Plot
plt.figure(figsize=(10,8))

plt.barh(
    feature_importance["Original_Feature"][::-1],
    feature_importance["Importance"][::-1]
)

plt.xlabel("Feature Importance")
plt.title("Aggregated Feature Importance - Random Forest")

plt.tight_layout()
plt.show()
```

### XGBoost

```python
from xgboost import XGBClassifier
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import LabelEncoder
```

```python
label_encoder = LabelEncoder()

y_train_encoded = label_encoder.fit_transform(y_train)
y_test_encoded = label_encoder.transform(y_test)

print(label_encoder.classes_)
```

```python
xgb_model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", XGBClassifier(
        n_estimators=200,
        max_depth=6,
        learning_rate=0.1,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42,
        eval_metric="logloss"
    ))
])
```

```python
xgb_model.fit(X_train, y_train_encoded)
```

```python
y_pred_xgb = xgb_model.predict(X_test)
```

```python
y_pred_labels = label_encoder.inverse_transform(y_pred_xgb)
```

```python
from sklearn.metrics import accuracy_score, classification_report

print(f"Accuracy: {accuracy_score(y_test, y_pred_labels):.4f}")

print("\nClassification Report:")
print(classification_report(y_test, y_pred_labels))
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, y_pred_labels)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=label_encoder.classes_
)

disp.plot(cmap="Blues")
plt.title("Confusion Matrix - XGBoost")
plt.show()
```

```python
param_dist = {
    "classifier__n_estimators": [100, 200, 300],
    "classifier__max_depth": [3, 5, 7, 9],
    "classifier__learning_rate": [0.01, 0.05, 0.1, 0.2],
    "classifier__subsample": [0.8, 0.9, 1.0],
    "classifier__colsample_bytree": [0.8, 0.9, 1.0],
    "classifier__min_child_weight": [1, 3, 5]
}
```

```python
from sklearn.model_selection import RandomizedSearchCV

random_search = RandomizedSearchCV(
    estimator=xgb_model,
    param_distributions=param_dist,
    n_iter=2,
    cv=2,
    scoring="f1",
    random_state=42,
    n_jobs=1,
    error_score="raise",
    verbose=2
)

random_search.fit(X_train, y_train)
```

```python
print("Best Parameters:")
print(random_search.best_params_)

print("\nBest CV F1 Score:")
print(random_search.best_score_)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
)

best_model = random_search.best_estimator_

y_pred = best_model.predict(X_test)

print("Accuracy :", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall   :", recall_score(y_test, y_pred))
print("F1 Score :", f1_score(y_test, y_pred))
```
